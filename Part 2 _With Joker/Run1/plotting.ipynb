```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python

b_options = [0.0001, 0.0005, 0.001,0.002, 0.003, 0.004, 0.005,0.006, 0.007,0.008, 0.009, 0.01, 0.02,0.03]
```

```python
train_acc = []
val_acc = []
test_acc = []
for b in b_options: 
    file = 'test_results_b='+str(b)+'_part2_b='+str(b)+'_200_10_300_16_.npy'
    test_results = np.load(file)
    
    train_acc.append(test_results[1])
    val_acc.append(test_results[2])
    test_acc.append(test_results[4])

```

```python
bar_width = 0.2  # Width of each bar
plt.close()
x = np.arange(len(b_options))

# Plotting grouped bar plots
#plt.figure(figsize=(10, 6))
plt.figure(figsize=(10, 6))
plt.bar(x, train_acc, width=bar_width, color='skyblue', label='Training Accuracy')
plt.bar(x + bar_width, val_acc, width=bar_width, color='orange', label='Validation Accuracy')
plt.bar(x + 2 * bar_width, test_acc, width=bar_width, color='green', label='Test Accuracy')

plt.xlabel('Scale Values (b) ')
plt.ylabel('Accuracy')
plt.title('Accuracies for Each Scale')
plt.xticks(x + bar_width, b_options)  # Set x-axis labels to category names

plt.legend()
#plt.tight_layout()

plt.savefig('run1_bar_plot_accuracies_along_the_scale.png', bbox_inches='tight')
plt.show()
```
